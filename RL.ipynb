```python
import numpy as np

# Gridworld setup
rows, cols = 4, 4
n_states = rows * cols
terminal_states = [0, 15]
gamma = 0.9
theta = 1e-4
max_iterations = 20  # limit to avoid infinite loop

# Actions: up, right, down, left
actions = [(-1, 0), (0, 1), (1, 0), (0, -1)]
action_names = ['↑', '→', '↓', '←']

# Use a simple policy: always move right
policy = np.ones(n_states, dtype=int)  # index 1 = '→'
V = np.zeros(n_states)

# Helper function to get next state
def step(state, action):
    if state in terminal_states:
        return state, 0
    row, col = divmod(state, cols)
    drow, dcol = actions[action]
    new_row, new_col = row + drow, col + dcol
    if 0 <= new_row < rows and 0 <= new_col < cols:
        next_state = new_row * cols + new_col
    else:
        next_state = state
    return next_state, -1

# Verbose policy evaluation
def policy_evaluation_verbose(policy, V, max_iter=20):
    for iteration in range(max_iter):
        print(f"\n=== Iteration {iteration} ===")
        delta = 0
        new_V = V.copy()
        for s in range(n_states):
            if s in terminal_states:
                print(f"State {s}: Terminal (V={V[s]:.2f})")
                continue
            a = policy[s]
            s_prime, r = step(s, a)
            old_v = V[s]
            new_v = r + gamma * V[s_prime]
            new_V[s] = new_v
            delta = max(delta, abs(old_v - new_v))
            print(f"State {s}: V = {r} + {gamma} * V({s_prime}) = {r} + {V[s_prime]:.2f} = {new_v:.2f}")
        V = new_V
        if delta < theta:
            print("\nConverged!")
            break
    return V

# Run verbose evaluation
final_V = policy_evaluation_verbose(policy, V, max_iter=max_iterations)
final_V.reshape((4, 4))
```
